# Some questions
## 1. which team won the most matches?

```python
match_wins = df['match_winner'].value_counts()
sns.barplot(y = match_wins.index, x = match_wins.values, palette = 'rainbow')
plt.title("Most matches win by team")
```

### 2. Toss decision trends

```python
sns.countplot(x = df['toss_decision'])
```

```python
count = df[df['toss_winner'] == df['match_winner']]['match_id'].count()
percentage = (count * 100)/df.shape[0]  
percentage.round(2)
```

```python
# how do teams win run or wickets 
sns.countplot(x = df['won_by'])
```

 # Key player performances

```python
# 1. Most player of the match awards
count = df['player_of_the_match'].value_counts().head(10)
count
```

```python
sns.barplot(x = count.values, y = count.index, palette = 'rainbow')
plt.title("Top 10 players of the match")
```

```python
# Top Scorer
high = df.groupby('top_scorer')['highscore'].sum().sort_values(ascending=False).head(5)
high
```

```python
high.plot(kind = 'bar')
```

```python
#10 best bowling figures
df['highest_wickets'] = df['best_bowling_figure'].apply(lambda x :x.split('--')[0])
df['highest_wickets'] = df['highest_wickets'].astype(int)
top_bowlers = df.groupby("best_bowling")['highest_wickets'].sum().sort_values(ascending = False).head(10)
top_bowlers.plot(kind = 'bar')
```

```python
# Venuw Analysis
venue_count = df['venue'].value_counts()
venue_count
```

```python
sns.barplot(y = venue_count.index, x = venue_count.values, palette = 'rainbow')
```

```python
# Who won the highest margin by runs?
df[df['won_by'] == 'Runs'].sort_values(by = 'margin', ascending = False).head(1)[['match_winner','margin']]
```

```python
# Which player has the highest individual score?
df[df['highscore'] == df['highscore'].max()][['top_scorer','highscore']]
```

```python
# Which bowler had the best bowling figures?
df[df['highest_wickets'] == df['highest_wickets'].max()][['best_bowling', 'best_bowling_figure']]
```
