```python
import seaborn as sns
import pandas as pd 
import cufflinks as cf
from plotly.offline import iplot
```

```python
tips = sns.load_dataset('tips')
tips
```

```python
tips['total_bill'].iplot()
```

```python
import plotly.express as px

fig = px.line(tips, y='total_bill')
fig.show()
```

```python
fig = px.histogram(tips, x='day', y = 'total_bill')
fig.show()
```

```python
fig = px.scatter(tips, y='tip', x = 'total_bill')
fig.show()
```

```python
fig = px.box(tips, y='total_bill')
fig.show()
```
